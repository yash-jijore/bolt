```python

import numpy as np
import matplotlib.pyplot as plt

# Define the function and its derivative
def f(x):
    return (x + 3) ** 2

def derivative(x):
    return 2 * (x + 3)

# Initial values
x = 2.0
learning_rate = 0.1
iterations = 100

# Gradient Descent
for i in range(iterations):
    gradient = derivative(x)
    x = x - learning_rate * gradient

print("Local minimum x:", round(x, 4))
print("Minimum value y:", round(f(x), 4))

# Plot the function
x_values = np.linspace(-6, 4, 100)
y_values = f(x_values)

plt.plot(x_values, y_values)
plt.scatter(x, f(x))
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Gradient Descent")
plt.grid(True)
plt.show()
```
